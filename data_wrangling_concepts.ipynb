```python
# Exercise 1: Concatenate
# Concept: Concatenation is like stacking papers on top of each other. 
#     You can stack rows (axis=0) or columns (axis=1).
```

```python
import pandas as pd
import numpy as np
```

```python
df1 = pd.DataFrame([['a', 1], ['b', 2]], columns=['letter', 'number'])
df2 = pd.DataFrame([['c', 1], ['d', 2]], columns=['letter', 'number'])

print("DataFrame 1:")
print(df1)
print("\nDataFrame 2:")
print(df2)

# Concatenate and reset index
result = pd.concat([df1, df2], axis=0, ignore_index=True)
print("\nConcatenated DataFrame:")
print(result)
print(f"Index type: {type(result.index)}")
```

```python
# Exercise 2: Merge - Joining DataFrames
## Concept: Merging is like joining two tables in a database. 
## You specify which column(s) to match on.
```

```python
#df1

df1_dict = {
        'id': ['1', '2', '3', '4', '5'],
        'Feature1': ['A', 'C', 'E', 'G', 'I'],
        'Feature2': ['B', 'D', 'F', 'H', 'J']}

df1 = pd.DataFrame(df1_dict, columns = ['id', 'Feature1', 'Feature2'])

#df2
df2_dict = {
        'id': ['1', '2', '6', '7', '8'],
        'Feature1': ['K', 'M', 'O', 'Q', 'S'],
        'Feature2': ['L', 'N', 'P', 'R', 'T']}

df2 = pd.DataFrame(df2_dict, columns = ['id', 'Feature1', 'Feature2'])
```

```python
### Inner join
pd.merge(df1, df2, on='id', how='inner')
```

```python
### Outer Join: All ids from both tables
pd.merge(df1, df2, on='id', how='outer')
```

```python
# Types of Merges:

# inner: Only rows that match in both DataFrames
# outer: All rows from both DataFrames (fills missing with NaN)
# left: All rows from left DataFrame
# right: All rows from right DataFrame
```
