```python
# Exercise 1: Concatenate
# Concept: Concatenation is like stacking papers on top of each other. 
#     You can stack rows (axis=0) or columns (axis=1).
```

```python
import pandas as pd
import numpy as np
```

```python
df1 = pd.DataFrame([['a', 1], ['b', 2]], columns=['letter', 'number'])
df2 = pd.DataFrame([['c', 1], ['d', 2]], columns=['letter', 'number'])

print("DataFrame 1:")
print(df1)
print("\nDataFrame 2:")
print(df2)

# Concatenate and reset index
result = pd.concat([df1, df2], axis=0, ignore_index=True)
print("\nConcatenated DataFrame:")
print(result)
print(f"Index type: {type(result.index)}")
```

```python
# Exercise 2: Merge - Joining DataFrames
## Concept: Merging is like joining two tables in a database. 
## You specify which column(s) to match on.
```

```python
#df1

df1_dict = {
        'id': ['1', '2', '3', '4', '5'],
        'Feature1': ['A', 'C', 'E', 'G', 'I'],
        'Feature2': ['B', 'D', 'F', 'H', 'J']}

df1 = pd.DataFrame(df1_dict, columns = ['id', 'Feature1', 'Feature2'])

#df2
df2_dict = {
        'id': ['1', '2', '6', '7', '8'],
        'Feature1': ['K', 'M', 'O', 'Q', 'S'],
        'Feature2': ['L', 'N', 'P', 'R', 'T']}

df2 = pd.DataFrame(df2_dict, columns = ['id', 'Feature1', 'Feature2'])
```

```python
### Inner join
pd.merge(df1, df2, on='id', how='inner')
```

```python
### Outer Join: All ids from both tables
pd.merge(df1, df2, on='id', how='outer', suffixes=("_df1", "_df2"))
```

```python
# Types of Merges:

# inner: Only rows that match in both DataFrames
# outer: All rows from both DataFrames (fills missing with NaN)
# left: All rows from left DataFrame
# right: All rows from right DataFrame
```

```python
# Exercise 3: Merge with MultiIndex
## Concept: MultiIndex is like having multiple layers of organization - think of a filing cabinet with drawers (level 1) and folders within drawers (level 2).
```

```python
#generate days
all_dates = pd.date_range('2021-01-01', '2021-12-15')
business_dates = pd.bdate_range('2021-01-01', '2021-12-31')

#generate tickers
tickers = ['AAPL', 'FB', 'GE', 'AMZN', 'DAI']

#create indexs
index_alt = pd.MultiIndex.from_product([all_dates, tickers], names=['Date', 'Ticker'])
index = pd.MultiIndex.from_product([business_dates, tickers], names=['Date', 'Ticker'])

# create DFs
market_data = pd.DataFrame(index=index,
                        data=np.random.randn(len(index), 3),
                        columns=['Open','Close','Close_Adjusted'])

alternative_data = pd.DataFrame(index=index_alt,
                                data=np.random.randn(len(index_alt), 2),
                                columns=['Twitter','Reddit'])
```

```python
market_data, alternative_data
```

```python
market_data.merge(alternative_data, how='left', left_index=True, right_index=True).fillna(0)
```

```python
# Alternatively
market_data.join(alternative_data, how="left").fillna(0)
```
