```python
# Exercise 1: Concatenate
# Concept: Concatenation is like stacking papers on top of each other. 
#     You can stack rows (axis=0) or columns (axis=1).
```

```python
import pandas as pd
import numpy as np
```

```python
df1 = pd.DataFrame([['a', 1], ['b', 2]], columns=['letter', 'number'])
df2 = pd.DataFrame([['c', 1], ['d', 2]], columns=['letter', 'number'])

print("DataFrame 1:")
print(df1)
print("\nDataFrame 2:")
print(df2)

# Concatenate and reset index
result = pd.concat([df1, df2], axis=0, ignore_index=True)
print("\nConcatenated DataFrame:")
print(result)
print(f"Index type: {type(result.index)}")
```
