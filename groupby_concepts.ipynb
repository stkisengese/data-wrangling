```python
# Exercise 4: GroupBy Apply - Custom Functions on Groups
## Concept: GroupBy is like sorting items into buckets, then doing the same operation on each bucket.
```

```python
import pandas as pd
import numpy as np
```

```python
def winsorize(df, quantiles):
    """
    Replace extreme values with specified percentiles
    df: DataFrame to winsorize
    quantiles: [lower_percentile, upper_percentile] e.g., [0.05, 0.95]
    """
    lower_bound = np.percentile(df, quantiles[0] * 100)
    upper_bound = np.percentile(df, quantiles[1] * 100)
    return df.clip(lower=lower_bound, upper=upper_bound)
```

```python
# Test the function
df = pd.DataFrame(range(1,11), columns=['sequence']).T
print(winsorize(df, [0.20, 0.80]).to_markdown())
```

```python
# consider each value belongs to a group
groups = np.concatenate([np.ones(10), np.ones(10)+1,  np.ones(10)+2, np.ones(10)+3, np.ones(10)+4])

df = pd.DataFrame(data= zip(groups,
                            range(1,51)),
                columns=["group", "sequence"])
```

```python
print(groups)
```

```python
df.T
```

```python
result = df.groupby("group")["sequence"].apply(winsorize, [0.05, 0.95])
result = result.reset_index(drop=True)
print(result.head(11).to_markdown())
```

```python
# Exercise 5: GroupBy Aggregation - Multiple Statistics
## Concept: Aggregation is like getting a summary report for each group - min, max, average, etc.
```

```python
# Common Aggregation Functions:

# sum(), mean(), median()
# min(), max(), std(), var()
# count(), nunique()
# first(), last()
```

```python
df_products = pd.DataFrame({
    'value': [20.45, 22.89, 32.12, 111.22, 33.22, 100, 99.99],
    'product': ['table', 'chair', 'chair', 'mobile phone', 'table', 'mobile phone', 'table']
})
df_products
```

```python
# Mutiple aggregations in one line
df_products.groupby('product')['value'].agg(['min', 'max', 'mean'])
```

```python
# Alternative method with custom column names
agg_dict = df_products.groupby('product').agg({
    'value': ['min', 'max', 'mean']
}).round(2)
agg_dict
```
